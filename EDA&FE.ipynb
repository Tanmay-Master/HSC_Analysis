```python
import pandas as pd
import os
import re

# List of your files (adjust the folder path if needed)
files = [
    "Data/HSC2021.xlsx",
    "Data/HSC2022.xlsx",
    "Data/HSC2023.xlsx",
    "Data/HSC2024.xlsx",
    "Data/HSC2025.xlsx"
]

# Empty list to hold DataFrames
dfs = []

for file in files:
    # Extract year from filename using regex (looks for 4 digits)
    match = re.search(r'(\d{4})', file)
    if match:
        year = int(match.group(1))
    else:
        raise ValueError(f"Year not found in filename: {file}")
    
    # Read the Excel file
    df = pd.read_excel(file)
    
    # Add the year column
    df['Year'] = year
    
    # Append to list
    dfs.append(df)

# Concatenate all DataFrames
combined_df = pd.concat(dfs, ignore_index=True)

# Save to CSV (without index, and ensure commas in college names are quoted)
combined_df.to_csv('combined_hsc_data.csv', index=False)

print("CSV file created: combined_hsc_data.csv")
print(f"Total rows: {len(combined_df)}")
```

# College Comparison

```python
#'Top Colleges by Average Pass % from 2021 to 2025' 
college_avg = df.groupby('Name of the college')['Pass Percent'].mean().reset_index()
college_avg = college_avg.sort_values('Pass Percent', ascending=False)

fig = px.bar(college_avg, 
             x='Pass Percent', y='Name of the college',
             orientation='h',
             title='Top Colleges by Average Pass % from 2021 to 2025',
             color='Pass Percent',
             color_continuous_scale='Viridis')
fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```

```python
stream_yearly = df.groupby(['Year', 'Stream'])['Candidates Registerd'].sum().reset_index()

fig = px.bar(stream_yearly, x='Year', y='Candidates Registerd', color='Stream',
             title='Total Registrations per Stream Over Years',
             barmode='stack')
fig.show()
```

```python
pass_stream = df.groupby(['Year', 'Stream'])['Pass Percent'].mean().reset_index()

fig = px.line(pass_stream, x='Year', y='Pass Percent', color='Stream',
              markers=True, title='Average Pass % by Stream')
fig.show()
```

```python
# Filter: SCIENCE and latest year
latest_year = df['Year'].max()
science_df = df[(df['Stream'] == 'SCIENCE') & (df['Year'] == latest_year)]

# Sort by Pass Percent
science_df = science_df.sort_values('Pass Percent', ascending=False)

fig = px.bar(science_df, 
             x='Pass Percent', 
             y='Name of the college',
             orientation='h',
             title=f'Science Stream – Pass % ({latest_year})',
             color='Pass Percent',
             color_continuous_scale='Blues',
             height=max(600, len(science_df)*20))   # dynamic height
fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```

```python
# Set the year you want to analyse
year = 2024          # <-- change this to any year, e.g., 2021, 2022, ...

# Filter: SCIENCE and the chosen year
science_df = df[(df['Stream'] == 'SCIENCE') & (df['Year'] == year)].copy()

# Create a formatted label for the bar text (e.g., "98.5%")
science_df['Pass_Label'] = science_df['Pass Percent'].apply(lambda x: f"{x:.1f}%")

# Sort for clearer ranking
science_df = science_df.sort_values('Pass Percent', ascending=False)

# Horizontal bar chart with labels
fig = px.bar(science_df,
             x='Pass Percent',
             y='Name of the college',
             orientation='h',
             text='Pass_Label',                       # <-- shows the percentage on the bar
             title=f'Science Stream – Pass % ({year})',
             color='Pass Percent',
             color_continuous_scale='Blues',
             height=max(600, len(science_df)*20))

# Fine-tune label appearance
fig.update_traces(textposition='outside', textfont_size=12)

# Widen x-axis a little so labels don't get clipped
fig.update_xaxes(range=[0, 105])   # assumes pass % is ≤100

fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("combined_hsc_data.csv")
df['Year'] = df['Year'].astype(int)

# ------- CHANGE THESE TWO VALUES -------
year   = 2024
stream = 'ARTS'
# ---------------------------------------

filtered = df[(df['Year'] == year) & (df['Stream'] == stream)].copy()
filtered['Pass_Label'] = filtered['Pass Percent'].apply(lambda x: f"{x:.1f}%")
filtered = filtered.sort_values('Pass Percent', ascending=False)

fig = px.bar(
    filtered,
    x='Pass Percent',
    y='Name of the college',
    orientation='h',
    text='Pass_Label',
    title=f'{stream} Stream – Pass % ({year})',
    color='Pass Percent',
    color_continuous_scale='Blues',
    height=max(400, len(filtered)*20)
)
fig.update_traces(textposition='outside', textfont_size=12)
fig.update_xaxes(range=[0, 105])
fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```
